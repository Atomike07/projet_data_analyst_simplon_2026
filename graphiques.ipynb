```
!pip install plotly -q

import pandas as pd
import plotly.express as px
import sqlite3

# Données
data = {
    'date': ['2022-01-01','2022-01-01','2022-01-02','2022-01-02','2022-01-03',
             '2022-01-03','2022-01-04','2022-01-04','2022-01-05','2022-01-05',
             '2022-01-06','2022-01-06','2022-01-07','2022-01-07','2022-01-08',
             '2022-01-08','2022-01-09','2022-01-09','2022-01-10','2022-01-10',
             '2022-01-11','2022-01-11','2022-01-12','2022-01-12','2022-01-13',
             '2022-01-13','2022-01-14','2022-01-14','2022-01-15','2022-01-15',
             '2022-01-16','2022-01-16','2022-01-17','2022-01-17','2022-01-18',
             '2022-01-18','2022-01-19','2022-01-19','2022-01-20','2022-01-20'],
    'produit': ['Produit A','Produit B','Produit A','Produit C','Produit B',
                'Produit C','Produit B','Produit C','Produit A','Produit B',
                'Produit C','Produit A','Produit B','Produit C','Produit A',
                'Produit B','Produit C','Produit A','Produit B','Produit C',
                'Produit A','Produit C','Produit A','Produit B','Produit C',
                'Produit A','Produit B','Produit C','Produit A','Produit B',
                'Produit C','Produit A','Produit B','Produit C','Produit A',
                'Produit B','Produit C','Produit B','Produit A','Produit B'],
    'prix': [10,15,10,20,15,20,15,20,10,15,20,10,15,20,10,15,20,10,15,20,
             10,20,10,15,20,10,15,20,10,15,20,10,15,20,10,15,20,15,10,15],
    'qte': [100,50,75,30,150,75,85,45,125,100,25,110,60,40,80,90,35,120,
            70,60,150,55,200,100,60,95,50,65,75,90,30,180,100,50,150,70,
            35,125,125,120],
    'region': ['Nord','Nord','Sud','Nord','Sud','Nord','Sud','Nord','Nord','Sud',
               'Nord','Nord','Nord','Sud','Nord','Sud','Nord','Sud','Nord','Sud',
               'Nord','Nord','Sud','Nord','Sud','Nord','Sud','Nord','Sud','Nord',
               'Nord','Sud','Sud','Nord','Sud','Nord','Sud','Nord','Nord','Sud']
}

df = pd.DataFrame(data)

# Base SQLite
conn = sqlite3.connect(":memory:")
df.to_sql("ventes", conn, index=False, if_exists="replace")

# Agrégations
ventes_produit = pd.read_sql_query("""
    SELECT produit, SUM(qte) AS total_qte,
           ROUND(SUM(prix * qte), 2) AS chiffre_affaires
    FROM ventes GROUP BY produit ORDER BY chiffre_affaires DESC;
""", conn)

ventes_region = pd.read_sql_query("""
    SELECT region, SUM(qte) AS total_qte,
           ROUND(SUM(prix * qte), 2) AS chiffre_affaires
    FROM ventes GROUP BY region ORDER BY chiffre_affaires DESC;
""", conn)

# Graphique exemple : ventes par région
fig0 = px.bar(ventes_region, x="region", y="total_qte",
              title="Ventes par région (quantités)",
              color="region", text="total_qte")
fig0.update_layout(showlegend=False)
fig0.show()

# Graphique 1 : ventes par produit
fig1 = px.bar(ventes_produit, x="produit", y="total_qte",
              title="Ventes par produit (quantités vendues)",
              color="produit", text="total_qte")
fig1.update_layout(showlegend=False)
fig1.show()

# Graphique 2 : CA par produit
fig2 = px.pie(ventes_produit, names="produit", values="chiffre_affaires",
              title="Chiffre d'affaires par produit (€)", hole=0.3)
fig2.update_traces(textinfo="label+percent+value")
fig2.show()

conn.close()
print("Terminé !")
```
